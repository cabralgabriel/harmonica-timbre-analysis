```python
import random
import pandas as pd

# Bloco A: configuracao de fabrica
# Bloco B: placas trocadas

configuracoes = {
    'A': [
        {'harmonica': 'SunRise',
         'comb': 'madeira',
         'key': 'C'},
        {'harmonica': 'TingHarp',
         'comb': 'plastico',
         'key': 'Bb'}
    ],
    'B': [
        {'harmonica': 'SunRise',
         'comb': 'plastico',
         'key': 'C'},
        {'harmonica': 'TingHarp',
         'comb': 'madeira',
         'key': 'Bb'}
    ]
}

metodos_por_nota = {
    'C': {
        'E4': 'natural',  # blow
        'F4': 'bend',     # draw bend
        'G4': 'natural',  # blow
        'A4': 'bend'      # draw bend
    },
    'Bb': {
        'E4': 'bend',     # draw bend
        'F4': 'natural',  # draw
        'G4': 'bend',     # draw bend
        'A4': 'natural'   # draw
    }
}

orificios_por_nota = {
    'C': {
        'E4': '2',
        'F4': "-2''",
        'G4': '3',
        'A4': "-3''"
    },
    'Bb': {
        'E4': "-2'",
        'F4': '-2',
        'G4': "-3''",
        'A4': '-3'
    }
}

random.seed(20260601)

ordem_gravacoes = []

for bloco in configuracoes:
    lista_bloco = []

    for config in configuracoes[bloco]:
        key = config['key']

        for nota in ['E4', 'F4', 'G4', 'A4']:
            metodo = metodos_por_nota[key][nota]
            orificio = orificios_por_nota[key][nota]

            # Adiciona todas combinações do bloco
            for rep in range(1, 6):
                lista_bloco.append({
                    'bloco': bloco,
                    'harmonica': config['harmonica'],
                    'comb': config['comb'],
                    'key': key,
                    'note': nota,
                    'metodo': metodo,
                    'orificio': orificio
                })

    random.shuffle(lista_bloco) # Aleatorizar dentro do bloco
    ordem_gravacoes.extend(lista_bloco)


df = pd.DataFrame(ordem_gravacoes)
df['rep'] = df.groupby(['bloco', 'harmonica', 'note']).cumcount() + 1
df = df[['bloco', 'key', 'comb', 'metodo', 'note', 'orificio', 'rep']]
print(df.to_string())

#df.to_excel('ordem_aleatorizada.xlsx', index=False)
```
