```python
import random
import pandas as pd

# block A: factory configuration
# block B: swapped plates

configurations = {
    'A': [
        {'harmonica': 'SunRise',
         'comb': 'wood',
         'key': 'C'},
        {'harmonica': 'TingHarp',
         'comb': 'abs',
         'key': 'Bb'}
    ],
    'B': [
        {'harmonica': 'SunRise',
         'comb': 'abs',
         'key': 'C'},
        {'harmonica': 'TingHarp',
         'comb': 'wood',
         'key': 'Bb'}
    ]
}

methods_per_note = {
    'C': {
        'E4': 'natural',  # blow
        'F4': 'bend',     # draw bend
        'G4': 'natural',  # blow
        'A4': 'bend'      # draw bend
    },
    'Bb': {
        'E4': 'bend',     # draw bend
        'F4': 'natural',  # draw
        'G4': 'bend',     # draw bend
        'A4': 'natural'   # draw
    }
}

holes_per_note = {
    'C': {
        'E4': '2',
        'F4': "-2''",
        'G4': '3',
        'A4': "-3''"
    },
    'Bb': {
        'E4': "-2'",
        'F4': '-2',
        'G4': "-3''",
        'A4': '-3'
    }
}

random.seed(20260601)

recording_order = []

for block in configurations:
    block_list = []

    for config in configurations[block]:
        key = config['key']

        for note in ['E4', 'F4', 'G4', 'A4']:
            method = methods_per_note[key][note]
            holes = holes_per_note[key][note]

            # adds all combinations of the block
            for rep in range(1, 6):
                block_list.append({
                    'block': block,
                    'harmonica': config['harmonica'],
                    'comb': config['comb'],
                    'key': key,
                    'note': note,
                    'method': method,
                    'holes': holes
                })

    random.shuffle(block_list) # shuffle within the block
    recording_order.extend(block_list)


df = pd.DataFrame(recording_order)
df['rep'] = df.groupby(['block', 'harmonica', 'note']).cumcount() + 1
df = df[['block', 'key', 'comb', 'method', 'note', 'holes', 'rep']]
print(df.to_string())

#df.to_excel('recording_order.xlsx', index=False)
```
